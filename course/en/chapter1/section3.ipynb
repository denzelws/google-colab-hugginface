```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

encoded_input = tokenizer("Hi, my name is Denzel")

print("=== Frase decodificada ===")
print(tokenizer.decode(encoded_input["input_ids"]))
print()

print("=== Dicionário completo ===")
print(encoded_input)
print()

print("=== Separação token por token ===")
tokens = tokenizer.convert_ids_to_tokens(encoded_input["input_ids"])
for id_, token in zip(encoded_input["input_ids"], tokens):
    print(f"{id_:>6} → {token}")
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

print("=== Cenário 1: múltiplas frases (varia attention_mask) ===")
sentences = [
    "Hi, my name is Denzel",
    "I love learning about transformers",
]
encoded_batch = tokenizer(sentences, padding=True)
print(encoded_batch)
print()

print("=== Cenário 2: par de frases relacionadas (varia token_type_ids) ===")
question = "Where do I work?"
context = "My name is Denzel and I work at Hugging Face"
encoded_pair = tokenizer(question, context)
print(encoded_pair)
print()

print("=== Decodificando o par, pra ver onde cada segmento começa ===")
print(tokenizer.decode(encoded_pair["input_ids"]))
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

print("=== Cenário 2: par de frases relacionadas (varia token_type_ids) ===")
question = "Where do I work?"
context = "My name is Denzel and I work at Hugging Face"
encoded_pair = tokenizer(question, context)
print(encoded_pair)
print()

print("=== Tokens do par completo, com token_type_id de cada um ===")
tokens = tokenizer.convert_ids_to_tokens(encoded_pair["input_ids"])
for id_, token, type_id in zip(encoded_pair["input_ids"], tokens, encoded_pair["token_type_ids"]):
    segmento = "question" if type_id == 0 else "context"
    marcador = "  ← 'work'" if token.lower() == "work" else ""
    print(f"{id_:>6} → {token:<15} (segmento: {segmento}){marcador}")
print()

print("=== Verificação isolada: 'work' tokenizado sozinho ===")
work_id = tokenizer.convert_tokens_to_ids("work")
print(f"ID do token 'work' no vocabulário: {work_id}")
```

```python
from transformers import AutoTokenizer

checkpoints = {
    "BERT (bert-base-cased)": "bert-base-cased",
    "GPT-2 (gpt2)": "gpt2",
    "T5 (t5-small)": "t5-small",
}

for label, checkpoint in checkpoints.items():
    tokenizer = AutoTokenizer.from_pretrained(checkpoint)
    algorithm = type(tokenizer.backend_tokenizer.model)
    print(f"{label:<25} → {algorithm}")
```
